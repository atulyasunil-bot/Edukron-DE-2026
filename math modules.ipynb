```python
# 1. Write a program to calculate the square root of a number using the `math` module.
import math
number = 25
square_root = math.sqrt(number)
print(f"The square root of {number} is {square_root}")
```

```python
# 2. Find the cube root of a given number using mathematical operations.
number = 27
cube_root = number ** (1/3)
print(f"The cube root of {number} is {cube_root}")
```

```python
# 3. Calculate the factorial of a number using `math.factorial()`.
number = 5
factorial = math.factorial(number)
print(f"The factorial of {number} is {factorial}")
```

```python
# 4. Find the GCD (Greatest Common Divisor) of two numbers using `math.gcd()`.
number1 = 12
number2 = 18
gcd = math.gcd(number1, number2)
print(f"The GCD of {number1} and {number2} is {gcd}")
```

```python
# 5.Find the LCM of two numbers using `math.lcm().
number1 = 12
number2 = 18
lcm = math.lcm(number1,number2)
print(f"The LCM of {number1} and {number2} is {lcm}")
```

```python
# 6.Calculate the value of 5 raised to the power of 8 using `math.pow()`.
import math
result = math.pow(5, 8)
print(result)
```

```python
# 7. Find the absolute value of a negative decimal number using `math.fabs()`.
number = -7.5
result = math.fabs(number)
print(result)
```

```python
# 8. Round a floating-point number up using `math.ceil()`.
import math
number = 3.2
result = math.ceil(number)
print(result)
```

```python
# 9. Round a floating-point number down using `math.floor()`.
import math
number = 3.8
result = math.floor(number)
print(result)
```

```python
# 10. Round a floating-point number to the nearest integer using `round()`.
number = 3.6
result = round(number)
print(result)
```

```python
# 11. Find the remainder of two numbers using `math.fmod()`. 
number1 = 10.5
number2 = 3.2
result = math.fmod(number1, number2)
print(result)
```

```python
# 12. Find the quotient and remainder using `divmod()`.
number1 = 10
number2 = 3
quotient, remainder = divmod(number1, number2)
print(f"The quotient of {number1} and {number2} is {quotient} and the remainder is {remainder}")
```

```python
# 13. Calculate the sine, cosine, and tangent of 45 degrees.
angle = math.radians(45)
print(f"sin(45°) = {math.sin(angle)}")
print(f"cos(45°) = {math.cos(angle)}")
print(f"tan(45°) = {math.tan(angle)}")
```

```python
# 14. Convert an angle from degrees to radians.
angle_degrees = 45
angle_radians = math.radians(angle_degrees)
print(f"{angle_degrees} degrees is equal to {angle_radians} radians.")
```

```python
# 15. Convert an angle from radians to degrees.
angle_radians = math.pi / 4
angle_degrees = math.degrees(angle_radians)
print(f"{angle_radians} radians is equal to {angle_degrees} degrees.")
```

```python
# 16. Find the logarithm (base 10) of a number using `math.log10()`.
number = 100
result = math.log10(number)
print(f"The logarithm (base 10) of {number} is {result}")
```

```python
# 17. Find the natural logarithm of a number using `math.log()`.
number = 10
result = math.log(number)
print(f"The natural logarithm of {number} is {result}")
```

```python
# 18. Calculate e raised to the power of x using `math.exp()`.
x = 2
result = math.exp(x)
print(f"e raised to the power of {x} is {result}")
```

```python
# 19. Find the value of π (`pi`) and Euler's number (`e`) from the `math` module.
import math
pi_value = math.pi
e_value = math.e
print(f"The value of π is {pi_value}")
print(f"The value of Euler's number is {e_value}")
```

```python
# 20. Find the distance between two points using `math.dist()`.
point1 = (2, 3)
point2 = (5, 7)
distance = math.dist(point1, point2)
print(f"The distance between points {point1} and {point2} is {distance}")
```

```python
# 21. Calculate the hypotenuse of a right triangle using `math.hypot()`.
a = 3
b = 4
hypotenuse = math.hypot(a, b)
print(f"The hypotenuse of the right triangle with sides {a} and {b} is {hypotenuse}")
```

```python
# 22. Find the combination (nCr) using `math.comb()`.
n = 5
r = 2
combination = math.comb(n, r)
print(f"The combination of {n} choose {r} is {combination}")
```

```python
#23. Find the permutation (nPr) using `math.perm()`.
n = 5
r = 2
permutation = math.perm(n, r)
print(f"The permutation of {n} choose {r} is {permutation}")
```

```python
# 24. Check whether a number is finite, infinite, or NaN using `math.isfinite()`, `math.isinf()`, and `math.isnan()`.
x = 5.0
print(f"Is {x} finite? {math.isfinite(x)}")
print(f"Is {x} infinite? {math.isinf(x)}")
print(f"Is {x} NaN? {math.isnan(x)}")
```

```python
# 25. Find the product of all numbers in a list using `math.prod()`.
numbers = [2, 3, 4, 5]
product = math.prod(numbers)
print(f"The product of all numbers in the list {numbers} is {product}")
```

```python
# 26. Find the sum of square roots of all numbers in a list.
numbers = [1, 4, 9, 16]
sum_of_square_roots = sum(math.hypot(x, 0) for x in numbers)
print(f"The sum of square roots of all numbers in the list {numbers} is {sum_of_square_roots}")
```

```python
# 27. Find the largest integer less than or equal to the square root of a given number.
number = 20
largest_integer = math.floor(math.sqrt(number))
print(f"The largest integer less than or equal to the square root of {number} is {largest_integer}")
```

```python
# 28. Calculate compound interest using `math.pow()`.
principal = 1000 
rate = 0.05         
time = 2         
compound_interest = principal * math.pow((1 + rate), time)
print(f"The compound interest after {time} years is {compound_interest}")
```

```python
# 29. Find the area and circumference of a circle using `math.pi`.
radius = 5
area = math.pi * radius ** 2
circumference = 2 * math.pi * radius
print(f"The area of the circle with radius {radius} is {area}")
print(f"The circumference of the circle with radius {radius} is {circumference}")
```

```python
# 30. Create a scientific calculator that supports square root, factorial, logarithm, power, trigonometric functions, GCD, and LCM.
print("=== Simple Scientific Calculator ===")
print("1. Square Root")
print("2. Factorial")
print("3. Logarithm (base 10)")
print("4. Power (x^y)")
print("5. Sine (sin)")
print("6. Cosine (cos)")
print("7. Tangent (tan)")
print("8. GCD")
print("9. LCM")

choice = input("Enter choice (1-9): ")

if choice == "1":
    num = float(input("Enter a number: "))
    result = math.sqrt(num)
    print("Result:", result)

elif choice == "2":
    num = int(input("Enter an integer: "))
    result = math.factorial(num)
    print("Result:", result)

elif choice == "3":
    num = float(input("Enter a number: "))
    result = math.log10(num)
    print("Result:", result)

elif choice == "4":
    base = float(input("Enter base: "))
    exp = float(input("Enter exponent: "))
    result = math.pow(base, exp)
    print("Result:", result)

elif choice == "5":
    angle = float(input("Enter angle in degrees: "))
    radians = math.radians(angle)
    print("Result:", math.sin(radians))

elif choice == "6":
    angle = float(input("Enter angle in degrees: "))
    radians = math.radians(angle)
    print("Result:", math.cos(radians))

elif choice == "7":
    angle = float(input("Enter angle in degrees: "))
    radians = math.radians(angle)
    print("Result:", math.tan(radians))

elif choice == "8":
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))
    print("GCD is:", math.gcd(num1, num2))

elif choice == "9":
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))
    print("LCM is:", math.lcm(num1, num2))

else:
    print("Invalid choice!")
```
