## Table Relationships

```
application_train / application_test
        | SK_ID_CURR
        |----------------------------------------------|
        v                                               v
     bureau.csv                                   POS_CASH_balance.csv
        | SK_ID_BUREAU                                  | SK_ID_PREV
        v
   bureau_balance.csv
```

- `SK_ID_CURR` links every table back to a single Home Credit customer/application.
- `SK_ID_BUREAU` links `bureau.csv` to its monthly history in `bureau_balance.csv`.
- `SK_ID_PREV` links previous/POS-CASH/installment/credit-card tables to a specific past loan.

Note: `previous_application.csv`, `installments_payments.csv`, and `credit_card_balance.csv` were not part of
this project's uploaded file set; the Streamlit app pages that use them detect this automatically and show
a placeholder with instructions instead of failing.
