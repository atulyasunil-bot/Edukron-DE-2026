````python
Sample Sets
bank_branches = {"MG Road", "Indiranagar", "Whitefield", "Electronic City"}

customers = {"Rahul", "Anita", "Kiran", "Sneha", "Vikram"}

new_customers = {"Amit", "Priya", "Rahul", "Divya"}

loan_customers = {"Rahul", "Kiran", "Asha", "Ramesh"}

savings_customers = {"Rahul", "Sneha", "Priya", "Nikhil"}

transaction_ids = {101, 102, 103, 104, 105}

balances = {85000, 120000, 45000, 95000, 275000}

cities = {"Bangalore", "Hyderabad", "Chennai", "Pune"}

new_cities = {"Mumbai", "Bangalore", "Delhi"}

account_types = {"Savings", "Current", "Loan"}
````

````python

# Print the customers set.
print(customers)
#Find the number of elements in the customers set.
print(len(customers))
#Check whether "Rahul" exists in the customers set.
print("Rahul" in customers)
#Check whether "Meena" exists in the customers set.
print("Meena" in customers)
#Iterate through all elements in the bank_branches set.
for branch in bank_branches:
    print(branch)

add()
#Add "Meena" to the customers set.
customers.add("Meena")
print(customers)
#Add "Mysore" to the cities set.
cities.add("Mysore")
print(cities)
#Add "Savings" to the account_types set.
account_types.add("Savings")
print(account_types)

update()
#Update the customers set with new_customers.
customers.update(new_customers)
print(customers)
#Update the cities set with new_cities.
cities.update(new_cities)
print(cities)

remove()
#Remove "Rahul" from the customers set.
customers.remove("Rahul")
print(customers)
#Remove "Bangalore" from the cities set.
cities.remove("Bangalore")
print(cities)
#Remove "Pune" from the cities set.
cities.remove("Pune")
print(cities)

discard()
#Discard "Anita" from the customers set.
customers.discard("Anita")
print(customers)
#Discard "Delhi" from the cities set.
cities.discard("Delhi")
print(cities)
#Discard "Kolkata" from the cities set.
cities.discard("Kolkata")
print(cities)

pop()
#Remove a random element from the transaction_ids set using pop().
transaction_ids.pop()
print(transaction_ids)
#Remove a random element from the balances set using pop().
balances.pop()
print(balances)
#Remove another random element from the account_types set.
account_types.pop()
print(account_types)

clear()
#Clear the loan_customers set.
loan_customers.clear()
print(loan_customers)
#Clear the savings_customers set.
savings_customers.clear()
print(savings_customers)
#Remove all elements from the balances set.
balances.clear()
print(balances)

copy()
#Create a copy of the customers set.
customers_copy = customers.copy()
print(customers_copy)
#Create a copy of the cities set.
cities_copy = cities.copy()
print(cities_copy)
#Modify the copied set and verify that the original set remains unchanged.
cities_copy.add("Ahmedabad")
print(cities_copy)
print(cities)

union()
#Find the union of customers and new_customers.
print(customers.union(new_customers))
#Find the union of cities and new_cities.
print(cities.union(new_cities))

intersection()
#Find the common customers between customers and new_customers.
print(customers.intersection(new_customers))
#Find the common cities between cities and new_cities.
print(cities.intersection(new_cities))

difference()
#Find customers who have both Savings and Loan accounts.
print(customers.intersection(account_types))
#Find customers present only in customers but not in new_customers.
print(customers.difference(new_customers))
#Find customers who have only Savings accounts.
print(customers.difference(account_types))
#Find customers that are present in either customers or new_customers but not both.
print(customers.symmetric_difference(new_customers))

issubset()
#Check whether {"Rahul", "Kiran"} is a subset of loan_customers.
print({"Rahul", "Kiran"}.issubset(loan_customers))
#Check whether {"Rahul", "Anita"} is a subset of customers.
print({"Rahul", "Anita"}.issubset(customers))

issuperset()
#Check whether customers is a superset of {"Rahul", "Anita"}.
print(customers.issuperset({"Rahul", "Anita"}))
#Check whether new_cities is a superset of cities.
print(new_cities.issuperset(cities))

isdisjoint()
#Check whether loan_customers and new_cities are disjoint.
print(loan_customers.isdisjoint(new_cities))
#Check whether customers and new_customers are disjoint.
print(customers.isdisjoint(new_customers))

Miscellaneous
#Find the maximum balance in the balances set.
print(max(balances))
#Find the minimum balance in the balances set.
print(min(balances))
#Find the total of all balances.
print(sum(balances))
#Sort the balances set in ascending order.
print(sorted(balances))
#Sort the balances set in descending order.
print(sorted(balances, reverse=True))
#Convert the customers set into a list.
print(list(customers))
#Convert the cities set into a tuple.
print(tuple(cities))
#Remove duplicates from the list: ["Rahul", "Rahul", "Anita", "Sneha", "Sneha"] by converting it into a set.
print(set(["Rahul", "Rahul", "Anita", "Sneha", "Sneha"]))
#Create an empty set.
account_types = set()
print(account_types)
#Add five employee names to the empty set.
account_types.add("Rahul")
account_types.add("Anita")
account_types.add("Kiran")
account_types.add("Sneha")
account_types.add("Vikram")
print(account_types)
#Find the length of the account_types set.
print(len(account_types))
#Check whether "Rahul" exists in the account_types set.
print("Rahul" in account_types)
#Check whether "Meena" exists in the account_types set.
print("Meena" in account_types)
#Print every element of the loan_customers set using a loop.
for customer in loan_customers:
    print(customer)
#Add a new customer "Meena" to the loan_customers set.
loan_customers.add("Meena")
print(loan_customers)
#Clear the transaction_ids set and verify that it is empty.
transaction_ids.clear()
print(transaction_ids)
````
