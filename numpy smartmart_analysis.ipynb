# SmartMart Retail Analytics — NumPy Assignment

## Business Scenario

You are working as a Data Analyst for **SmartMart Retail Stores**. The company operates five stores in different cities and sells six major product categories.

Management wants to analyze sales performance, expenses, profits, targets, discounts, inventory, and store-level performance using NumPy.

Two main NumPy arrays are provided:
1. `sales_data` — monthly sales revenue generated by each store for different product categories.
2. `cost_data` — total cost incurred by each store for the same product categories.


## Part 1: Dataset Creation

```python
import numpy as np

# Store names
stores = np.array([
    "Bengaluru",
    "Hyderabad",
    "Chennai",
    "Mumbai",
    "Pune"
])

# Product categories
categories = np.array([
    "Electronics",
    "Groceries",
    "Clothing",
    "Furniture",
    "Beauty",
    "Sports"
])

# Sales revenue in rupees (rows = stores, columns = categories)
sales_data = np.array([
    [850000, 420000, 310000, 275000, 190000, 225000],
    [780000, 460000, 295000, 310000, 210000, 240000],
    [920000, 390000, 340000, 260000, 185000, 215000],
    [1050000, 510000, 375000, 345000, 250000, 290000],
    [690000, 405000, 280000, 225000, 175000, 205000]
])

# Cost incurred in rupees (rows = stores, columns = categories)
cost_data = np.array([
    [640000, 315000, 220000, 205000, 125000, 160000],
    [590000, 350000, 210000, 225000, 140000, 170000],
    [700000, 295000, 245000, 195000, 120000, 155000],
    [790000, 390000, 265000, 250000, 165000, 205000],
    [530000, 310000, 205000, 170000, 115000, 145000]
])

print("Stores:")
print(stores)

print("\nProduct Categories:")
print(categories)

print("\nSales Data:")
print(sales_data)

print("\nCost Data:")
print(cost_data)
```

Display number of dimensions, shape, size, dtype, and memory usage of `sales_data`.

```python
# Number of dimensions in sales_data
print("Number of dimensions:", sales_data.ndim)

# Shape of sales_data -> (stores, categories)
print("Shape of sales_data:", sales_data.shape)

# Total number of values
print("Total number of values:", sales_data.size)

# Data type of the array
print("Data type of the array:", sales_data.dtype)

# Memory occupied by the array (in bytes)
print("Memory occupied (bytes):", sales_data.nbytes)
```

```python
# Total number of stores
num_stores = stores.shape[0]

# Total number of product categories
num_categories = categories.shape[0]

# Total number of store-category combinations
num_combinations = sales_data.size

print("Total number of stores:", num_stores)
print("Total number of product categories:", num_categories)
print("Total number of store-category combinations:", num_combinations)
```

## Part 10: Discount Scenario

```python
discount_rates = np.array([
    0.10,  # Electronics
    0.05,  # Groceries
    0.15,  # Clothing
    0.12,  # Furniture
    0.08,  # Beauty
    0.10   # Sports
])
print("Discount rates by category:")
for cat, val in zip(categories, discount_rates):
    print(f"{cat}: {val*100:.0f}%")
```

### Question 58: Discount Amount Matrix

```python
# Broadcasting discount_rates (6,) across sales_data (5,6)
discount_amount = sales_data * discount_rates
print("Discount Amount Matrix:")
print(np.round(discount_amount, 2))
```

### Question 59: Sales After Discount

```python
sales_after_discount = sales_data - discount_amount
print("Sales After Discount:")
print(np.round(sales_after_discount, 2))
```

### Question 60: Profit After Discount

```python
profit_after_discount = sales_after_discount - cost_data
print("Profit After Discount:")
print(np.round(profit_after_discount, 2))
```

### Question 61: Profit Margin After Discount

```python
profit_margin_after_discount = np.round((profit_after_discount / sales_after_discount) * 100, 2)
print("Profit Margin After Discount (%):")
print(profit_margin_after_discount)
```

### Question 62: Loss-Making Transactions After Discount

```python
loss_mask = profit_after_discount < 0
store_rows, cat_cols = np.where(loss_mask)

if len(store_rows) == 0:
    print("No store-category combination becomes loss-making after discounts.")
else:
    print("Loss-making combinations after discount:")
    for r, c in zip(store_rows, cat_cols):
        print(f"Store: {stores[r]}, Category: {categories[c]}, Profit: {profit_after_discount[r,c]:.2f}")
```

### Question 63: Discount Impact

### Question 64: Store Most Affected by Discounts

### Question 91: Missing Value Detection

### Question 92: Replace Missing Values with Zero

### Question 93: Replace with Company Average

```python
company_avg = np.nanmean(sales_with_missing)
sales_filled_avg = np.where(np.isnan(sales_with_missing), company_avg, sales_with_missing)

print("Company average (ignoring NaNs):", round(company_avg, 2))
print("Sales data with missing values replaced by company average:")
print(sales_filled_avg)
```

### Question 94: Replace with Category Average

```python
category_avg = np.nanmean(sales_with_missing, axis=0)

# Broadcast category averages into a full array matching shape, then use where
category_avg_broadcast = np.tile(category_avg, (sales_with_missing.shape[0], 1))
sales_filled_cat_avg = np.where(np.isnan(sales_with_missing), category_avg_broadcast, sales_with_missing)

print("Category averages (ignoring NaNs):", np.round(category_avg, 2))
print("Sales data with missing values replaced by category average:")
print(sales_filled_cat_avg)
```

### Question 95: Missing Value Impact

```python
total_before = np.nansum(sales_with_missing)
total_after_zero = sales_filled_zero.sum()

print("Total sales ignoring missing values (nansum):", total_before)
print("Total sales after replacing missing values with zero:", total_after_zero)
print("Difference:", total_before - total_after_zero)
```
