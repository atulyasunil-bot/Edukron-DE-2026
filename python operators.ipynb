```python
#1. Write a program to perform addition, subtraction, multiplication, division, floor division, modulus, and exponentiation on two numbers.
import math
num1 = 5.0
num2 = 3.0

# Perform calculations
addition = num1 + num2
subtraction = num1 - num2
multiplication = num1 * num2
division = num1 / num2
floor_division = num1 // num2
modulus = num1 % num2
exponentiation = num1 ** num2

print("\n--- Results ---")
print(f"Addition ({num1} + {num2}):", addition)
print(f"Subtraction ({num1} - {num2}):", subtraction)
print(f"Multiplication ({num1} * {num2}):", multiplication)
print(f"Division ({num1} / {num2}):", division)
print(f"Floor Division ({num1} // {num2}):", floor_division)
print(f"Modulus ({num1} % {num2}):", modulus)
print(f"Exponentiation ({num1} ** {num2}):", exponentiation)
```

```python
# 2. Swap two numbers without using a third variable.
num1 = 4
num2 = 6
# Swapping without a third variable
num1 = num1 + num2
num2 = num1 - num2
num1 = num1 - num2
print(f"After swapping: num1 = {num1}, num2 = {num2}")
```

```python
# 3. Find whether a number is even or odd using the modulus operator.
number = 7
if number % 2 == 0:
    print(f"{number} is even.")
else:
    print(f"{number} is odd.")
```

```python
# 4. Reverse a three-digit number using arithmetic operators.
number = 678
if 100 <= number <= 999:
    reversed_number = (number % 10) * 100 + ((number // 10) % 10) * 10 + (number // 100)
    print(f"Reversed number: {reversed_number}")
else:
    print("Please enter a valid three-digit number.")
```

```python
# 5. Find the last digit and first digit of a given number.
number = 8875
last_digit = number % 10
first_digit = number // (10 ** (len(str(number)) - 1))
print(f"Last digit: {last_digit}")
print(f"First digit: {first_digit}")
```

```python
# 6. Demonstrate all assignment operators (`=`, `+=`, `-=`, `*=`, `/=`, `//=`, `%=` and `**=`) on a variable.
x = 10
print(f"Initial value: {x}")
x += 5
print(f"After += 5: {x}")
x -= 3
print(f"After -= 3: {x}")
x *= 2
print(f"After *= 2: {x}")
x /= 4
print(f"After /= 4: {x}")
x //= 2
print(f"After //= 2: {x}")
x %= 3
print(f"After %= 3: {x}")
x **= 2
print(f"After **= 2: {x}")
```

```python
# 7. Update a bank account balance using compound assignment operators.
balance = 1000
print(f"Initial balance: {balance:.2f}")
balance += 500
print(f"After deposit of 500: {balance:.2f}")
balance -= 200
print(f"After withdrawal of 200: {balance:.2f}")
```

```python
# 8. Calculate the total marks using `+=` inside a loop.
total_marks = 0
for i in range(5):
    marks = int(input(f"Enter marks for subject {i+1}: "))
    total_marks += marks
print(f"Total marks: {total_marks}")
```

```python
# 11. Find the largest among three numbers using comparison operators.
num1 = 7
num2 = 10
num3 = 5
if num1 >= num2 and num1 >= num3:
    largest = num1
elif num2 >= num1 and num2 >= num3:
    largest = num2
else:
    largest = num3
print(f"The largest number is: {largest}")
```

```python
# 17. Determine whether a year is a leap year using logical conditions.
year = 2020
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} is a leap year.")
else:
    print(f"{year} is not a leap year.")
```

```python
# 18. Find the bitwise AND, OR, XOR of two numbers.
num1 = 5  # Binary: 0101
num2 = 3  # Binary: 0011
bitwise_and = num1 & num2
bitwise_or = num1 | num2
bitwise_xor = num1 ^ num2
print(f"Bitwise AND of {num1} and {num2}: {bitwise_and}")
print(f"Bitwise OR of {num1} and {num2}: {bitwise_or}")
print(f"Bitwise XOR of {num1} and {num2}: {bitwise_xor}")
```

```python
# 19. Perform left shift and right shift operations.
num = 5  # Binary: 0101       
left_shift = num << 1   # Binary: 1010               
right_shift = num >> 1  # Binary: 0010
print(f"Left shift of {num}: {left_shift}")
print(f"Right shift of {num}: {right_shift}")
```

```python
# 20. Check whether a number is even using a bitwise operator.
number = 8
if number & 1 == 0:
    print(f"{number} is even.")
```

```python
# 21. Swap two numbers using the XOR operator.
num1 = 5
num2 = 3
# Swapping using XOR
num1 = num1 ^ num2
num2 = num1 ^ num2
num1 = num1 ^ num2
print(f"After swapping using XOR: num1 = {num1}, num2 = {num2}")
```

```python
# 28. Build a simple calculator using arithmetic and comparison operators.
def calculator():
    print("Simple Calculator")
    print("Select operation:")
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")
    
    choice = input("Enter choice (1/2/3/4): ")
    
    if choice in ['1', '2', '3', '4']:
        # Get inputs from the user instead of hardcoding
        num1 = float(input("Enter first number: "))
        num2 = float(input("Enter second number: "))
        
        if choice == '1':
            result = num1 + num2
            operation = "Addition"
        elif choice == '2':
            result = num1 - num2
            operation = "Subtraction"
        elif choice == '3':
            result = num1 * num2
            operation = "Multiplication"
        elif choice == '4':
            if num2 != 0:
                result = num1 / num2
                operation = "Division"
            else:
                print("Error: Division by zero is not allowed.")
                return
        
        print(f"{operation} of {num1} and {num2} is: {result}")
    else:
        print("Invalid input")

# Run the function
calculator()
```

```python
# 29. Create a grading system using comparison and logical operators.
def grading_system(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    else:
        return "D"
    # Example usage
score = 85
grade = grading_system(score)
print(f"Score: {score}, Grade: {grade}")
```

```python
# 30. Build a menu-driven calculator that uses arithmetic, comparison, logical, assignment, membership, identity, and bitwise operators wherever applicable.
def menu_driven_calculator():
    while True:
        print("\nMenu-Driven Calculator")
        print("1. Arithmetic Operations")
        print("2. Comparison Operations")
        print("3. Logical Operations")
        print("4. Assignment Operations")
        print("5. Membership Operations")
        print("6. Identity Operations")
        print("7. Bitwise Operations")
        print("8. Exit")

        choice = input("Enter your choice (1-8): ")

        if choice == '1':
            num1 = float(input("Enter first number: "))
            num2 = float(input("Enter second number: "))
            print(f"Addition: {num1 + num2}")
            print(f"Subtraction: {num1 - num2}")
            print(f"Multiplication: {num1 * num2}")
            if num2 != 0:
                print(f"Division: {num1 / num2}")
            else:
                print("Division by zero is not allowed.")
            print(f"Modulus: {num1 % num2}")
            print(f"Exponentiation: {num1 ** num2}")

        elif choice == '2':
            num1 = float(input("Enter first number: "))
            num2 = float(input("Enter second number: "))
            print(f"{num1} > {num2}: {num1 > num2}")
            print(f"{num1} < {num2}: {num1 < num2}")
            print(f"{num1} == {num2}: {num1 == num2}")

        elif choice == '3':
            a = bool(int(input("Enter first boolean (0 or 1): ")))
            b = bool(int(input("Enter second boolean (0 or 1): ")))
            print(f"{a} AND {b}: {a and b}")
            print(f"{a} OR {b}: {a or b}")
            print(f"NOT {a}: {not a}")

        elif choice == '4':
            x = int(input("Enter a number for assignment operations: "))
            x += 5
            print(f"x after += 5: {x}")
            x -= 3
            print(f"x after -= 3: {x}")

        elif choice == '5':
            string = input("Enter a string: ")
            char = input("Enter a character to check membership: ")
            print(f"'{char}' in '{string}': {char in string}")

        elif choice == '6':
            a = input("Enter first object: ")
            b = input("Enter second object: ")
            print(f"{a} is {b}: {a is b}")

        elif choice == '7':
            num1 = int(input("Enter first number: "))
            num2 = int(input("Enter second number: "))
            print(f"{num1} & {num2}: {num1 & num2}")
            print(f"{num1} | {num2}: {num1 | num2}")
            print(f"{num1} ^ {num2}: {num1 ^ num2}")
            print(f"~{num1}: {~num1}")
            print(f"{num1} << 1: {num1 << 1}")
            print(f"{num1} >> 1: {num1 >> 1}")

        elif choice == '8':
            print("Exiting the calculator.")
            break

        else:
            print("Invalid choice. Please enter a number between 1 and 8.")

# Run the menu-driven calculator
menu_driven_calculator()
```
