```python
def analyze_text(text):
  vowels="aeiouAEIOU"
  characters_count=len(text)
  words_count=len(text.split())
  vowels_count=0
  digits_count=0
  space_count=0
  for char in text:
    if char in vowels:
      vowels_count+=1
    if char.isdigit():
      digits_count+=1
    if char == " ":
      space_count+=1
  return {
      "total_characters":characters_count,
      "total_words":words_count,
      "total_vowels":vowels_count,
      "total_digits":digits_count,
      "total_spaces":space_count
  }
text=str(input("enter your sentence:"))
result=analyze_text(text)
print(f"Total characters: {result['total_characters']}")
print(f"Total words: {result['total_words']}")
print(f"Total vowels: {result['total_vowels']}")
print(f"Total digits: {result['total_digits']}")
print(f"Total spaces: {result['total_spaces']}")
```

Assignment 92: Closure-Based Bank Balance
Write a function named create_bank_account().

Requirements
Store the balance in the outer function.
Define inner functions for:
Deposit
Withdrawal
Balance enquiry
Use nonlocal to update the balance.
Reject withdrawals greater than the available balance.
Return all three inner functions.
Hint
Return multiple functions:

return deposit, withdraw, check_balance

```python
def create_bank_account(initial_balance=0):
    balance = initial_balance

    def deposit(amount):
        nonlocal balance
        if amount <= 0:
            print("Deposit amount must be positive")
        else:
            balance += amount
            print(f"Deposited {amount}. New balance: {balance}")

    def withdraw(amount):
        nonlocal balance
        if amount <= 0:
            print("Withdrawal amount must be positive")
        elif amount > balance:
            print(f"Insufficient balance. Available: {balance}")
        else:
            balance -= amount
            print(f"Withdrew {amount}. New balance: {balance}")

    def check_balance():
        print(f"Current balance: {balance}")
        return balance

    return deposit, withdraw, check_balance


deposit, withdraw, check_balance = create_bank_account(1000)

check_balance()
deposit(500)
withdraw(300)
withdraw(5000)
check_balance()
```

Part 12: Menu-Driven Program
Create a menu containing the following options:

```python
from functools import reduce

company_name = "Edukron"  # Global scope

employees = [
    {"id": 101, "name": "Ravi", "department": "IT", "salary": 60000, "performance_score": 88, "experience": 5},
    {"id": 102, "name": "Anitha", "department": "HR", "salary": 45000, "performance_score": 72, "experience": 4},
    {"id": 103, "name": "Kiran", "department": "IT", "salary": 75000, "performance_score": 94, "experience": 7},
    {"id": 104, "name": "Meena", "department": "Finance", "salary": 55000, "performance_score": 81, "experience": 6},
    {"id": 105, "name": "Arjun", "department": "IT", "salary": 40000, "performance_score": 65, "experience": 2}
]


# ---------- Display ----------
def display_employees(emp_list):
    if not emp_list:
        print("No employees found.")
        return
    print(f"{'ID':<5}{'Name':<10}{'Department':<12}{'Salary':<10}{'Score':<8}{'Experience':<10}")
    print("-" * 55)
    for emp in emp_list:
        print(f"{emp['id']:<5}{emp['name']:<10}{emp['department']:<12}"
              f"{emp['salary']:<10}{emp['performance_score']:<8}{emp['experience']:<10}")
    print()


# ---------- Part 2: Names ----------
def get_employee_names(emp_list):
    return list(map(lambda emp: emp["name"], emp_list))


# ---------- Part 3: High Performers ----------
def get_high_performers(emp_list, threshold=80):
    return list(filter(lambda emp: emp["performance_score"] >= threshold, emp_list))


# ---------- Part 4: Filter by Department ----------
def filter_by_department(department):
    return list(filter(lambda emp: emp["department"] == department, employees))


# ---------- Part 5: Revised Salaries ----------
def calculate_increment(emp):
    score = emp["performance_score"]
    if score >= 90:
        percent = 15
    elif score >= 80:
        percent = 10
    elif score >= 70:
        percent = 5
    else:
        percent = 0
    new_salary = round(emp["salary"] * (1 + percent / 100))
    return {**emp, "salary": new_salary}


def get_revised_employees(emp_list):
    return list(map(calculate_increment, emp_list))


# ---------- Part 6: Total Salary Expense ----------
def total_salary(emp_list):
    return reduce(lambda acc, emp: acc + emp["salary"], emp_list, 0)


def salary_expense_report(emp_list):
    revised = get_revised_employees(emp_list)
    current_total = total_salary(emp_list)
    revised_total = total_salary(revised)
    additional = revised_total - current_total
    return current_total, revised_total, additional


# ---------- Part 7: Highest-Paid ----------
def get_highest_paid(emp_list):
    if not emp_list:
        return None
    return reduce(lambda a, b: a if a["salary"] >= b["salary"] else b, emp_list)


# ---------- Part 8: Sorting ----------
def sort_employees(emp_list, key_name):
    return sorted(emp_list, key=lambda emp: emp[key_name])


# ---------- Part 9 & 11: Nested Report with LEGB ----------
def generate_department_report(department):
    dept_name = department  # Enclosing scope
    dept_employees = filter_by_department(dept_name)

    def employee_count():
        return len(dept_employees)

    def dept_total_salary():
        return reduce(lambda acc, emp: acc + emp["salary"], dept_employees, 0)

    def average_salary():
        count = employee_count()
        return dept_total_salary() / count if count else 0

    def average_score():
        count = employee_count()
        if count == 0:
            return 0
        total_score = reduce(lambda acc, emp: acc + emp["performance_score"], dept_employees, 0)
        return total_score / count

    def report_title():
        # 'title' -> Local scope (defined right here)
        # 'dept_name' -> Enclosing scope (from generate_department_report)
        # 'company_name' -> Global scope (module-level variable)
        title = f"{company_name} - {dept_name} Department Report"
        return title

    return {
        "title": report_title(),
        "department": department,
        "employee_count": employee_count(),
        "total_salary": dept_total_salary(),
        "average_salary": round(average_salary(), 2),
        "average_score": round(average_score(), 2)
    }


# ---------- Menu-Driven Program ----------
def print_menu():
    print("""
========== Employee Performance Analysis System ==========
1. Display all employees
2. Display employee names
3. Display high performers
4. Search by department
5. Calculate revised salaries
6. Calculate total salary expense
7. Find highest-paid employee
8. Sort employees
9. Generate department report
10. Exit
============================================================""")


def main():
    while True:
        print_menu()
        choice = input("Enter your choice (1-10): ").strip()

        if choice == "1":
            display_employees(employees)

        elif choice == "2":
            print(get_employee_names(employees))

        elif choice == "3":
            try:
                threshold = int(input("Enter minimum performance score (default 80): ") or 80)
            except ValueError:
                threshold = 80
            display_employees(get_high_performers(employees, threshold))

        elif choice == "4":
            dept = input("Enter department name (IT/HR/Finance): ").strip()
            display_employees(filter_by_department(dept))

        elif choice == "5":
            revised = get_revised_employees(employees)
            print("Revised Salaries:")
            display_employees(revised)

        elif choice == "6":
            current_total, revised_total, additional = salary_expense_report(employees)
            print("Current Total Salary:", current_total)
            print("Revised Total Salary:", revised_total)
            print("Total Additional Expense:", additional)

        elif choice == "7":
            top_emp = get_highest_paid(employees)
            print("Highest-Paid Employee:")
            print(top_emp)

        elif choice == "8":
            print("Sort by: name / salary / performance_score / experience")
            key_name = input("Enter sort field: ").strip()
            valid_keys = {"name", "salary", "performance_score", "experience"}
            if key_name in valid_keys:
                display_employees(sort_employees(employees, key_name))
            else:
                print("Invalid sort field.")

        elif choice == "9":
            dept = input("Enter department name (IT/HR/Finance): ").strip()
            report = generate_department_report(dept)
            print(report["title"])
            print(f"Employee Count : {report['employee_count']}")
            print(f"Total Salary   : {report['total_salary']}")
            print(f"Average Salary : {report['average_salary']}")
            print(f"Average Score  : {report['average_score']}")

        elif choice == "10":
            print("Exiting program. Goodbye!")
            break

        else:
            print("Invalid choice. Please enter a number between 1 and 10.")


if __name__ == "__main__":
    main()
```
